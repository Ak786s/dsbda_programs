```python
# -------------------------------
# IMPORT LIBRARIES
# -------------------------------
import nltk
from sklearn.feature_extraction.text import TfidfVectorizer

# -------------------------------
# DOWNLOAD REQUIRED DATA (RUN ONCE)
# -------------------------------
nltk.download('punkt')
nltk.download('punkt_tab')  # extra fix for tokenizer error
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag

# -------------------------------
# SAMPLE TEXT
# -------------------------------
text = "I am learning data science and I love coding. Data science is very interesting."

print("\nOriginal Text:\n", text)

# -------------------------------
# 1. TOKENIZATION (SAFE METHOD)
# -------------------------------
try:
    tokens = word_tokenize(text)
except:
    print("Tokenizer issue, using fallback...")
    tokens = text.split()

print("\nTokens:\n", tokens)

# -------------------------------
# 2. POS TAGGING
# -------------------------------
try:
    pos_tags = pos_tag(tokens)
    print("\nPOS Tags:\n", pos_tags)
except:
    print("\nPOS Tagging skipped (resource issue)")

# -------------------------------
# 3. STOP WORDS REMOVAL
# -------------------------------
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in tokens if word.lower() not in stop_words]

print("\nAfter Stop Words Removal:\n", filtered_words)

# -------------------------------
# 4. STEMMING
# -------------------------------
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]

print("\nStemmed Words:\n", stemmed_words)

# -------------------------------
# 5. LEMMATIZATION
# -------------------------------
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\nLemmatized Words:\n", lemmatized_words)

# -------------------------------
# 6. TF-IDF
# -------------------------------
documents = [
    "I am learning data science",
    "data science is interesting",
    "machine learning is powerful"
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents)

print("\nFeature Names:\n", vectorizer.get_feature_names_out())
print("\nTF-IDF Matrix:\n", X.toarray())
```
