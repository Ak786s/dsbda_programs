```python
import pandas as pd

# Create dataset
data = {
    'AgeGroup': ['Young', 'Young', 'Adult', 'Adult', 'Senior'],
    'Income': [20000, 25000, 40000, 45000, 30000]
}

df = pd.DataFrame(data)

# Group by AgeGroup and calculate stats
result = df.groupby('AgeGroup')['Income'].agg(['mean', 'median', 'min', 'max', 'std'])

print(result)
```
